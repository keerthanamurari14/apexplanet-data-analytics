```python
from pathlib import Path
import pandas as pd

# Find the project folder
current_path = Path.cwd()
project_root = current_path if (current_path / "data").exists() else current_path.parent

# Load the cleaned Superstore dataset
csv_path = project_root / "data" / "processed" / "superstore_cleaned.csv"
df = pd.read_csv(csv_path)

# Display the dataset details
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())

# Preview the first five rows
df.head()
```

### Findings

```python
# Query 4: Total sales by product category
query4 = """
SELECT
    category,
    SUM(sales) AS total_sales,
    AVG(sales) AS average_sales,
    COUNT(*) AS total_orders
FROM superstore
GROUP BY category
ORDER BY total_sales DESC;
"""

# Query 5: Categories with total sales greater than 100000
query5 = """
SELECT
    category,
    SUM(sales) AS total_sales
FROM superstore
GROUP BY category
HAVING SUM(sales) > 100000
ORDER BY total_sales DESC;
"""

# Query 6: Overall sales and profit statistics
query6 = """
SELECT
    COUNT(*) AS total_records,
    SUM(sales) AS total_sales,
    AVG(sales) AS average_sales,
    MIN(sales) AS minimum_sales,
    MAX(sales) AS maximum_sales,
    SUM(profit) AS total_profit
FROM superstore;
"""

print("Query 4: Sales by category")
display(pd.read_sql_query(query4, connection))

print("Query 5: Categories with sales above 100000")
display(pd.read_sql_query(query5, connection))

print("Query 6: Overall sales and profit statistics")
display(pd.read_sql_query(query6, connection))
```

### Findings

* Technology recorded the highest total sales, followed by Furniture and Office Supplies.
* Aggregate functions summarize sales, profit, and record counts.
* `HAVING` filters grouped results based on specified conditions.

```python
# Create two small tables for JOIN practice
connection.execute("DROP TABLE IF EXISTS customers_practice;")
connection.execute("DROP TABLE IF EXISTS orders_practice;")

connection.execute("""
CREATE TABLE customers_practice AS
SELECT DISTINCT customer_id, customer_name, segment
FROM superstore;
""")

connection.execute("""
CREATE TABLE orders_practice AS
SELECT order_id, customer_id, sales, profit
FROM superstore;
""")

connection.commit()

# Query 7: INNER JOIN
query7 = """
SELECT
    c.customer_name,
    c.segment,
    o.order_id,
    o.sales
FROM customers_practice AS c
INNER JOIN orders_practice AS o
    ON c.customer_id = o.customer_id
LIMIT 10;
"""

# Query 8: LEFT JOIN
query8 = """
SELECT
    c.customer_name,
    c.segment,
    o.order_id,
    o.sales
FROM customers_practice AS c
LEFT JOIN orders_practice AS o
    ON c.customer_id = o.customer_id
LIMIT 10;
"""

print("Query 7: INNER JOIN")
display(pd.read_sql_query(query7, connection))

print("Query 8: LEFT JOIN")
display(pd.read_sql_query(query8, connection))
```

**Finding:** 
INNER JOIN returns only records with matching values in both tables.
LEFT JOIN returns all customers, including those without matching orders.

```python
print("Products with sales above average:")

display(pd.read_sql_query("""
SELECT order_id, product_name, sales
FROM superstore
WHERE sales > (
    SELECT AVG(sales)
    FROM superstore
)
ORDER BY sales DESC
LIMIT 10;
""", connection))
```

**Finding:** The subquery identifies records with sales greater than the dataset's average sales.


```python
print("Category-wise sales using a CTE:")

display(pd.read_sql_query("""
WITH category_sales AS (
    SELECT
        category,
        SUM(sales) AS total_sales,
        AVG(sales) AS average_sales
    FROM superstore
    GROUP BY category
)
SELECT
    category,
    total_sales,
    average_sales
FROM category_sales
ORDER BY total_sales DESC;
""", connection))
```

**Finding:** Technology recorded the highest total sales, followed by Furniture and Office Supplies.

```python
print("Ranking orders by sales within each category:")

display(pd.read_sql_query("""
SELECT
    category,
    order_id,
    product_name,
    sales,
    ROW_NUMBER() OVER (
        PARTITION BY category
        ORDER BY sales DESC
    ) AS row_num
FROM superstore
ORDER BY category, row_num
LIMIT 15;
""", connection))
```

**Finding:** ROW_NUMBER() assigns a unique sequential number to each record within its category, ordered by sales.

```python
print("Ranking orders by sales within each category:")

display(pd.read_sql_query("""
SELECT
    category,
    order_id,
    product_name,
    sales,
    RANK() OVER (
        PARTITION BY category
        ORDER BY sales DESC
    ) AS sales_rank
FROM superstore
ORDER BY category, sales_rank
LIMIT 15;
""", connection))
```

**Finding:** RANK() assigns ranks based on sales within each category. Equal sales values receive the same rank.

```python
print("Comparing current sales with previous sales:")

display(pd.read_sql_query("""
SELECT
    order_date,
    order_id,
    sales,
    LAG(sales) OVER (
        ORDER BY order_date
    ) AS previous_sales
FROM superstore
ORDER BY order_date
LIMIT 10;
""", connection))
```

**Finding:** LAG() retrieves the previous row's sales value for comparison with the current row.

```python
print("Comparing current sales with next sales:")

display(pd.read_sql_query("""
SELECT
    order_date,
    order_id,
    sales,
    LEAD(sales) OVER (
        ORDER BY order_date
    ) AS next_sales
FROM superstore
ORDER BY order_date
LIMIT 10;
""", connection))
```

**Finding:** LEAD() retrieves the next row's sales value for comparison with the current row.

```python
print("Customer retention analysis:")

display(pd.read_sql_query("""
WITH customer_months AS (
    SELECT DISTINCT
        customer_id,
        strftime('%Y-%m', order_date) AS order_month
    FROM superstore
),
customer_activity AS (
    SELECT
        customer_id,
        COUNT(*) AS active_months
    FROM customer_months
    GROUP BY customer_id
)
SELECT
    COUNT(*) AS total_customers,
    SUM(
        CASE WHEN active_months > 1 THEN 1 ELSE 0 END
    ) AS repeat_month_customers,
    ROUND(
        100.0 * SUM(
            CASE WHEN active_months > 1 THEN 1 ELSE 0 END
        ) / COUNT(*), 2
    ) AS repeat_month_customer_percentage
FROM customer_activity;
""", connection))
```

### Findings

* The query identifies customers who placed orders in more than one month.
* The repeat-month customer percentage indicates how frequently customers return across different months.

```python
print("Monthly sales with a 3-month moving average:")

display(pd.read_sql_query("""
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', order_date) AS sales_month,
        SUM(sales) AS total_sales
    FROM superstore
    GROUP BY strftime('%Y-%m', order_date)
)
SELECT
    sales_month,
    total_sales,
    ROUND(
        AVG(total_sales) OVER (
            ORDER BY sales_month
            ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
        ), 2
    ) AS moving_average
FROM monthly_sales
ORDER BY sales_month;
""", connection))
```

### Findings

* The query calculates monthly sales and a 3-month moving average.
* The moving average smooths monthly fluctuations and helps identify sales trends.


```python
print("Cumulative monthly sales:")

display(pd.read_sql_query("""
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', order_date) AS sales_month,
        SUM(sales) AS total_sales
    FROM superstore
    GROUP BY strftime('%Y-%m', order_date)
)
SELECT
    sales_month,
    total_sales,
    SUM(total_sales) OVER (
        ORDER BY sales_month
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS cumulative_sales
FROM monthly_sales
ORDER BY sales_month;
""", connection))
```

### Findings

* The query calculates monthly sales and cumulative sales over time.
* Cumulative sales show how the total sales amount increases across successive months.
